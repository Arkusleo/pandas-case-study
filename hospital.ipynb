```python
hospital = pd.read_csv("Hospital_Patients.csv")
```

```python
print("First rows:\n", hospital.head())
print("Last rows:\n", hospital.tail())
print("Random sample:\n", hospital.sample(min(5, len(hospital)), random_state=7))
print("Shape:", hospital.shape)
print("Columns:", hospital.columns)
print("Data types:\n", hospital.dtypes)
print("Missing values:\n", hospital.isnull().sum())
```

```python
print("Numeric summary:\n", hospital.describe())
print("All-column summary:\n", hospital.describe(include="all"))
```

```python
stay = pd.to_numeric(hospital["StayDays"], errors="coerce")
```

```python
print("StayDays mean:", stay.mean())
print("StayDays median:", stay.median())
print("StayDays mode:\n", stay.mode())
print("StayDays minimum:", stay.min())
print("StayDays maximum:", stay.max())
print("StayDays standard deviation:", stay.std())
```

```python
bill_numeric = pd.to_numeric(hospital["BillAmount"], errors="coerce")
print("Bill mean:", bill_numeric.mean())
print("Bill median:", bill_numeric.median())
```

```python
print("Department counts:\n", hospital["Department"].value_counts())
print("Department percentages:\n", hospital["Department"].value_counts(normalize=True)*(100))
print("Insurance percentage:\n", hospital["Insurance"].value_counts(normalize=True)*(100))
print("Unique Gender values:", hospital["Gender"].nunique())
print("Gender values:", hospital["Gender"].unique())
```

```python
print("\nPART E - TASK 6")
dup_patient_mask = hospital.duplicated(keep=False)
print("Duplicate rows beyond first occurrences:", hospital.duplicated().sum())
print("All rows in duplicate groups:\n", hospital[dup_patient_mask])
hospital = hospital.drop_duplicates().copy()
print("PatientID unique:", hospital["PatientID"].is_unique)
print("Duplicate PatientID count:", hospital["PatientID"].duplicated().sum())
```

```python
hospital["Age"] = pd.to_numeric(hospital["Age"], errors="coerce")
hospital["Age"] = hospital["Age"].fillna(hospital["Age"].median()).astype(int)
print("Missing Age after conversion:", hospital["Age"].isnull().sum())
print(hospital["Age"].dtype)
```

```python
print(hospital[hospital["StayDays"] > 10])
print(hospital[hospital["SatisfactionScore"] < 5])

print(hospital[(hospital["Insurance"] == "No") & (hospital["BillAmount"] > 50000)])
```

```python

print(hospital.sort_values("AdmitDate", ascending=False).head(5))
print(hospital.nlargest(5, "BillAmount"))
```

```python
bq1 = hospital["BillAmount"].quantile(0.25)
bq3 = hospital["BillAmount"].quantile(0.75)
biqr = bq3 - bq1
bill_lower = bq1 - 1.5 * biqr
bill_upper = bq3 + 1.5 * biqr
bill_outliers = hospital[(hospital["BillAmount"] < bill_lower) | (hospital["BillAmount"] > bill_upper)]
```

```python
hospital_clean = hospital[(hospital["BillAmount"] >= bill_lower) & (hospital["BillAmount"] <= bill_upper)].copy()
hospital_clean["Stay_Type"] = hospital_clean["StayDays"].apply(
    lambda x: "Short" if x <= 3 else "Medium" if x <= 7 else "Long"
)


hospital_clean["Age_Group"] = hospital_clean["Age"].apply(
    lambda x: "Child" if x < 18 else "Adult" if x <= 59 else "Senior"
)

print(hospital_clean["Stay_Type"].value_counts())
print(hospital_clean["Age_Group"].value_counts())
```

```python
print(hospital_clean.groupby("Stay_Type")["SatisfactionScore"].mean())
print(hospital_clean.groupby("Age_Group")["SatisfactionScore"].mean())
```

```python
hospital_clean.to_csv("cleaned_hospital.csv", index=False)

df = pd.read_csv("cleaned_hospital.csv")

print(df.shape)
print(df.dtypes)
```
