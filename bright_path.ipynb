```python
import pandas as pd

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue_values = [42, 45, 50, 48, 55, 60, 58, 62, 65, 70, 75, 85]
expense_values = [38, 40, 42, 47, 50, 52, 55, 60, 61, 64, 66, 72]
```

```python
# A1: Create Revenue and Expenses Series with month labels
revenue = pd.Series(revenue_values, index=months, name="Revenue")
expenses = pd.Series(expense_values, index=months, name="Expenses")

print(revenue)
print(expenses)
```

```python
# A2: Python lists concatenate instead of adding element by element
a = [42, 45, 50]
b = [38, 40, 42]
print(a + b)
```

```python
# A3: Calculate profit using a for loop
profit_list = []
for i in range(len(revenue_values)):
    profit_list.append(revenue_values[i] - expense_values[i])

print(profit_list)
```

```python
# A4: Calculate profit using a Pandas Series expression and compare with A3
profit = revenue - expenses
profit.name = "Profit"
print(profit)
print( profit.tolist() == profit_list)
```

```python
# B1: Monthly profit
print(profit)
```

```python
# B2: Monthly profit margin (%)
margin = (profit / revenue * 100).round(1)
margin.name = "Margin (%)"
print(margin)
```

```python
# B3: Full-year totals and overall margin
total_revenue = revenue.sum()
total_expenses = expenses.sum()
total_profit = profit.sum()
overall_margin = round(total_profit / total_revenue * 100, 1)

print("Total Revenue:", total_revenue)
print("Total Expenses:", total_expenses)
print("Total Profit:", total_profit)
print("Overall Margin:", overall_margin)
```

```python
# C1: Quarterly profit using loc
Q1 = profit.loc["Jan":"Mar"]
Q2 = profit.loc["Apr":"Jun"]
Q3 = profit.loc["Jul":"Sep"]
Q4 = profit.loc["Oct":"Dec"]
print("Q1 Profit:", Q1.sum())
print("Q2 Profit:", Q2.sum())
print("Q3 Profit:", Q3.sum())
print("Q4 Profit:", Q4.sum())
quarter_profits = pd.Series({"Q1": Q1.sum(), "Q2": Q2.sum(),
                             "Q3": Q3.sum(), "Q4": Q4.sum()})
print("Strongest Quarter:", quarter_profits.idxmax())
```

```python
# C2: Half-yearly profit using iloc
H1 = profit.iloc[0:6]
H2 = profit.iloc[6:12]

print("H1 Profit:", H1.sum())
print("H2 Profit:", H2.sum())
```

```python
# C3: Profit for October, November and December
festive = profit.loc[["Oct", "Nov", "Dec"]]
print(festive)
```

```python
# C4: iloc equivalent of loc["Apr":"Jun"]
print(profit.iloc[3:6])
```

```python
# D1: Months where profit margin exceeds 10%
print(margin[margin > 10])
```

```python
# D2: Months where profit is below ₹4 lakh
print(profit[profit < 4])
```

```python
# D3: Best and worst months using Boolean conditions
best = profit[profit == profit.max()]
worst = profit[profit == profit.min()]

print("Best month and profit:",best)
print("Worst month and profit:",worst)
```

```python
# D4: Top 3 months by profit
print(profit.sort_values(ascending=False).head(3))
```

```python
# D5: Average monthly profit and months above average
average_profit = profit.mean()
print("Average monthly profit:", round(average_profit, 2))
print("Months above average:",profit[profit > average_profit])
```

```python
# E1: Classify each month's profit
def profit_status(p):
    if p >= 8:
        return "Strong"
    elif p >= 4:
        return "Stable"
    else:
        return "Weak"

status = profit.apply(profit_status)
print(status)
```

```python
# E3: Calculate tax: 25% of positive profit, otherwise 0
def tax_payable(p):
    if p > 0:
        return p * 0.25
    return 0
tax = profit.apply(tax_payable)
print(tax)
```

```python
# E4: Total tax and post-tax profit
total_tax = tax.sum()
post_tax_profit = total_profit - total_tax
print("Total Tax:", total_tax)
print("Post-tax Profit:", post_tax_profit)
```

```python
# F1: Remove December revenue and subtract expenses
rev_partial = revenue.iloc[0:11]
result = rev_partial - expenses
print(result)
```
