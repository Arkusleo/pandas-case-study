```python
# Task 9
print("Rows with missing Salary:\n", emp[emp["Salary"].isnull()])
emp["Salary"] = emp["Salary"].fillna(emp["Salary"].median())
emp["PerformanceScore"] = emp["PerformanceScore"].fillna(emp["PerformanceScore"].median())
print("Missing Salary after filling:", emp["Salary"].isnull().sum())
print("Missing PerformanceScore after filling:", emp["PerformanceScore"].isnull().sum())
```

```python
corr_exp_salary = emp["Experience"].corr(emp["Salary"])
corr_att_perf = emp["Attendance"].corr(emp["PerformanceScore"])
corr_age_exp = emp["Age"].corr(emp["Experience"])
print("Experience vs Salary:", corr_exp_salary)
print("Attendance vs PerformanceScore:", corr_att_perf)
print("Age vs Experience:", corr_age_exp)
```

```python
q1 = emp["Salary"].quantile(0.25)
q3 = emp["Salary"].quantile(0.75)
iqr = q3 - q1
salary_lower = q1 - 1.5 * iqr
salary_upper = q3 + 1.5 * iqr
salary_outliers = emp[(emp["Salary"] < salary_lower) | (emp["Salary"] > salary_upper)]
print("Salary outlier rows:\n", salary_outliers)
```

```python
emp_clean = emp[(emp["Salary"] >= salary_lower) & (emp["Salary"] <= salary_upper)].copy()
emp_clean["Overall_Score"] = emp_clean[["PerformanceScore", "Attendance"]].mean(axis=1)
emp_clean["Annual_Bonus"] = emp_clean["Salary"] * 0.10
print(emp_clean[["Name", "Overall_Score", "Annual_Bonus"]].head())
```

```python
emp_clean["Grade"] = emp_clean["PerformanceScore"].apply(
    lambda score: "Excellent" if score >= 85 else ("Good" if score >= 70 else "Needs Improvement")
)

print(emp_clean["Grade"])
```

```python
emp_clean.to_csv("cleaned_hr.csv", index=False)
emp_check = pd.read_csv("cleaned_hr.csv")
print("Exported and reloaded shape:", emp_check.shape)
print("Reloaded data types:\n", emp_check.dtypes)
```
